```python
secreto = 37
acertou = False
for i in range(5):
    numero = int(input("Digite seu palpite: "))
    if numero == secreto:
        print("Você acertou!")
        acertou = True
        break
    elif numero < secreto:
        print("O número secreto é maior")
    elif numero > secreto:
        print("O número secreto é menor")
if acertou == False:
    print("Você acabou as tentativas")
```

```python
saldo = 200
entradas = 0
retiradas = 0
recusadas = 0
total_entrada = 0
total_saida = 0

codigo = int(input("Digite o código da operação: "))

while codigo != 0:

    if codigo == 1:
        valor = float(input("Digite o valor da entrada: "))

        saldo = saldo + valor
        entradas = entradas + 1
        total_entrada = total_entrada + valor

    elif codigo == 2:
        valor = float(input("Digite o valor da retirada: "))

        if valor <= saldo:
            saldo = saldo - valor
            retiradas = retiradas + 1
            total_saida = total_saida + valor
        else:
            print("Saldo insuficiente")
            recusadas = recusadas + 1

    codigo = int(input("Digite o código da operação: "))

print("Saldo final:", saldo)
print("Quantidade de entradas:", entradas)
print("Quantidade de retiradas realizadas:", retiradas)
print("Quantidade de retiradas recusadas:", recusadas)
print("Total que entrou:", total_entrada)
print("Total que saiu:", total_saida)
```
