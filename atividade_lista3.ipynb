```python
notas = []
for i in range(5):
    nota = float(input("Digite uma nota: "))
    notas.append(nota)
media = 0
for i in range(5):
    media = media + notas[i]
media = media / 5
quantidade = 0
for i in range(5):
    if notas[i] >= media:
        quantidade = quantidade + 1
print("Notas:", notas)
print("Média:", media)
print("Quantidade de notas maiores ou iguais à média:", quantidade)
```

```python
x = int(input("Digite o valor de X: "))
y = int(input("Digite o valor de Y: "))
coordenada = (x, y)
if x == 0 and y == 0:
    print("A coordenada está na origem")
elif x == 0:
    print("A coordenada está no eixo Y")
elif y == 0:
    print("A coordenada está no eixo X")
else:
    if x > 0:
        if y > 0:
            print("A coordenada está no 1º quadrante")
        else:
            print("A coordenada está no 4º quadrante")
    else:
        if y > 0:
            print("A coordenada está no 2º quadrante")
        else:
            print("A coordenada está no 3º quadrante")
print("Coordenada:", coordenada)
```

```python
numeros = []
for i in range(8):
    numero = int(input("Digite um número: "))
    numeros.append(numero)
maior = numeros[0]
menor = numeros[0]
posicao_maior = 0
posicao_menor = 0
for i in range(1, 8):
    if numeros[i] > maior:
        maior = numeros[i]
        posicao_maior = i
    if numeros[i] < menor:
        menor = numeros[i]
        posicao_menor = i
print("Lista:", numeros)
print("Maior:", maior)
print("Posição do maior:", posicao_maior)
print("Menor:", menor)
print("Posição do menor:", posicao_menor)
```

```python
palavras = {}
for i in range(8):
    palavra = input("Digite uma palavra: ")
    if palavra in palavras:
        palavras[palavra] = palavras[palavra] + 1
    else:
        palavras[palavra] = 1
print("Frequência das palavras:")
for palavra in palavras:
    print(palavra, ":", palavras[palavra])
```

```python
estoque = {"arroz": 20,"feijao": 15,"macarrao": 10,"cafe": 8}
produto = input("Digite o produto vendido: ")
quantidade = int(input("Digite a quantidade vendida: "))
if produto in estoque:
    if quantidade <= estoque[produto]:
        estoque[produto] = estoque[produto] - quantidade
        print("Venda realizada")
    else:
        print("Estoque insuficiente")
else:
    print("Produto não encontrado")
print("Estoque final:", estoque)
```
