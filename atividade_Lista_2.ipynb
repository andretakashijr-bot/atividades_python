```python
primeiro = int(input("Digite o primeiro número: "))
maior = primeiro
menor = primeiro

for i in range(7):
  numero = int(input("Digite os números: "))
  if numero > maior:
    maior = numero
  if numero < menor:
    menor = numero
  print("O maior número é: ", maior)
```

```python
primeiro = int(input("Digite o primeiro número: "))
maior = primeiro
menor = primeiro

for i in range(7):
  numero = int(input("Digite os números: "))
  if numero > maior:
    menor = maior
    maior = numero
  if numero > menor and numero < maior:
    menor = numero
  print("O maior número é: ", maior)
  print("O segundo maior número é: ", menor)
```
